```python
import pandas as pd

def load_students_from_dataset():

    file_path = input("Enter CSV file path: ")

    try:
        df = pd.read_csv(file_path)
    except:
        print("Error: Could not read the CSV file.")
        return

    required_columns = [
        "student_id",
        "name",
        "roll_no",
        "branch",
        "section",
        "semester"
    ]

    # Check whether all required columns exist
    for column in required_columns:
        if column not in df.columns:
            print("Missing column:", column)
            return

    # Clear old student records
    students.clear()

    # Store dataset records in students dictionary
    for _, row in df.iterrows():

        student_id = str(row["student_id"])

        students[student_id] = {
            "name": row["name"],
            "roll_no": row["roll_no"],
            "branch": row["branch"],
            "section": row["section"],
            "semester": row["semester"]
        }

    print("\nStudent dataset loaded successfully!")
    print("Total students:", len(students))
```

```python
def generate_sitting_plan():

    if not students:
        print("No students available.")
        return

    if not rooms:
        print("No rooms available.")
        return

    if not exams:
        print("No exams available.")
        return



    exam_id = input("Enter Exam ID: ").strip()

    exam = find_exam(exam_id)

    if exam is None:
      print("Exam not found.")
      return
    detained_ids = set(detained_students.get(exam_id, {}).keys())
    eligible_students = {
    student_id: student
    for student_id, student in students.items()
    if student_id not in detained_ids
}

    if not eligible_students:
      print("No eligible students available for this exam.")
      return

    total_capacity = sum(
    room["rows"] * room["seats_per_row"]
    for room in rooms
)

    if total_capacity < len(eligible_students):
      print("Not enough room capacity!")
      print("Eligible students:", len(eligible_students))
      print("Available seats:", total_capacity)
      print("Please add more rooms or seats.")
      return

    sitting_plan.clear()

    print("Total registered students:", len(students))
    print("Detained students for this exam:", len(detained_ids))
    print("Eligible students:", len(eligible_students))

    section_students = {}

    for student_id, student in eligible_students.items():

        section = student["section"]

        if section not in section_students:
            section_students[section] = []

        section_students[section].append(student_id)

    sections = sorted(section_students.keys())


# Arrange students in a rotating section order
    ordered_students = []

    while any(section_students[section] for section in sections):
      for section in sections:

        if section_students[section]:
            student_id = section_students[section].pop(0)
            ordered_students.append((student_id, section))


# Allocate the ordered students to room seats
    student_index = 0

    for room in rooms:
      for row in range(1, room["rows"] + 1):

        for seat in range(1, room["seats_per_row"] + 1):

            if student_index >= len(ordered_students):
                break

            student_id, section = ordered_students[student_index]

            assignment = {
                "exam_id": exam_id,
                "student_id": student_id,
                "room_no": room["room_no"],
                "row": row,
                "seat": seat,
                "section": section
            }

            sitting_plan.append(assignment)
            student_index += 1

        if student_index >= len(ordered_students):
            break

      if student_index >= len(ordered_students):
        break


    remaining_students = 0

    for section in section_students:
        remaining_students += len(section_students[section])

    if remaining_students > 0:

        print("Sitting plan generated, but some students could not be allocated.")
        print("Students remaining:", remaining_students)

    else:
        print("Sitting plan generated successfully!")
```

```python

def pre_exam_validation_report():

    print("\n========== PRE-EXAM VALIDATION REPORT ==========")

    if not sitting_plan:
        print("No sitting plan available. Generate a plan first.")
        return
    exam_id = input("Enter Exam ID to validate: ").strip()
    exam = find_exam(exam_id)
    if not exam:
        print(f"Error: Exam ID {exam_id} not found.")
        return


    # 1. Check allocation against eligible students for this exam
    exam_assignments = [
        assignment for assignment in sitting_plan
        if assignment["exam_id"] == exam_id
    ]

    allocated_ids = [
        assignment["student_id"]
        for assignment in exam_assignments
    ]

    registered_ids = set(students.keys())
    allocated_set = set(allocated_ids)
    detained_ids = set(detained_students.get(exam_id, {}).keys())

    eligible_ids = registered_ids - detained_ids
    missing_students = eligible_ids - allocated_set
    unknown_students = allocated_set - registered_ids

    print("\n1. Student Allocation")
    print("Registered students:", len(registered_ids))
    print("Detained students:", len(detained_ids))
    print("Expected eligible students:", len(eligible_ids))
    print("Allocated assignments:", len(exam_assignments))
    print("Eligible students without seats:", len(missing_students))
    print("Unknown student IDs:", len(unknown_students))


    # 2. Check duplicate seat assignments for students
    duplicate_students = len(allocated_ids) - len(allocated_set)

    print("\n2. Duplicate Student Allocations")
    print("Duplicate assignments:", duplicate_students)

    # 3. Check room capacity
    print("\n3. Room Capacity")

    capacity_violations = []

    for room in rooms:
        room_no = str(room["room_no"]).strip()
        capacity = room["rows"] * room["seats_per_row"]

        room_assignments = [
            assignment for assignment in sitting_plan
            if str(assignment["room_no"]).strip() == room_no
        ]

        if len(room_assignments) > capacity:
            capacity_violations.append(room_no)

        print(
            f"Room {room_no}: "
            f"{len(room_assignments)}/{capacity} seats occupied"
        )

    print("Rooms exceeding capacity:", len(capacity_violations))

    # 4. Check duplicate physical seats
    seat_positions = [
        (
            str(assignment["room_no"]).strip(),
            assignment["row"],
            assignment["seat"]
        )
        for assignment in sitting_plan
    ]

    duplicate_seats = len(seat_positions) - len(set(seat_positions))

    print("\n4. Physical Seat Conflicts")
    print("Duplicate seat positions:", duplicate_seats)

        # 5. Detained Student Validation
    print("\n5. Detained Student Validation")

    detained_ids = set(detained_students.get(exam_id, {}).keys())

    allocated_ids = {
        assignment["student_id"]
        for assignment in sitting_plan
        if assignment["exam_id"] == exam_id
    }

    incorrectly_allocated = detained_ids.intersection(allocated_ids)

    eligible_count = len(students) - len(detained_ids)

    print("Registered students:", len(students))
    print("Detained students:", len(detained_ids))
    print("Expected eligible students:", eligible_count)
    print("Allocated students:", len(allocated_ids))
    print("Detained students incorrectly allocated:", len(incorrectly_allocated))

    if incorrectly_allocated:
        print("FAIL: Some detained students have seats.")
    else:
        print("PASS: No detained students have seats.")
    print("\n========== VALIDATION SUMMARY ==========")


    if (
        not missing_students
        and not unknown_students
        and duplicate_students == 0
        and not capacity_violations
        and duplicate_seats == 0
        and not incorrectly_allocated
        and len(allocated_set) == len(eligible_ids)
    ):

        print("PASS: No issues found in these validation checks.")
    else:
        print("WARNING: Issues found. Review the report above.")

    print("========================================")
```

```python
from collections import deque

waiting_queue = deque()
```

```python
def create_waiting_queue():

    waiting_queue.clear()

    allocated_ids = set()

    for assignment in sitting_plan:
        allocated_ids.add(assignment["student_id"])

    for student_id in students:

        if student_id not in allocated_ids:
            waiting_queue.append(student_id)

    print("\nWaiting Queue Created.")

    print("Students waiting:", len(waiting_queue))
```

```python
def view_waiting_queue():

    if not waiting_queue:
        print("Waiting queue is empty.")
        return

    print("\n================================")
    print("         WAITING QUEUE")
    print("================================")

    position = 1

    for student_id in waiting_queue:

        student = students[student_id]

        print(
            position,
            "→",
            student_id,
            "|",
            student["name"],
            "| Section:",
            student["section"]
        )

        position += 1
```

```python
def process_next_waiting_student():

    if not waiting_queue:
        print("Waiting queue is empty.")
        return

    student_id = waiting_queue.popleft()

    student = students[student_id]

    print("\nNext student in queue:")

    print("Student ID:", student_id)
    print("Name:", student["name"])
    print("Roll No:", student["roll_no"])
    print("Section:", student["section"])
```

```python

def clear_sitting_plan():

    if not sitting_plan:
        print("Sitting plan is already empty.")
        return

    # Save the current plan before clearing it
    undo_stack.append(sitting_plan.copy())

    sitting_plan.clear()
    waiting_queue.clear()

    print("Sitting plan cleared successfully.")
    print("You can restore it using the Undo feature.")
```

```python
import pandas as pd
from google.colab import files

# Store detention records separately for each exam
detained_students = {}


def load_detained_students():
    global detained_students

    uploaded = files.upload()

    if not uploaded:
        print("No file uploaded.")
        return

    filename = next(iter(uploaded))

    if not filename.lower().endswith(".csv"):
        print("Please upload a CSV file.")
        return

    try:
        df = pd.read_csv(filename, dtype=str).fillna("")
    except Exception as error:
        print("Could not read the CSV file:", error)
        return

    required_columns = {"exam_id", "student_id", "reason"}

    if not required_columns.issubset(df.columns):
        print("Invalid CSV format.")
        print("Required columns: exam_id, student_id, reason")
        return

    valid_records = {}
    invalid_count = 0

    for _, row in df.iterrows():
        exam_id = row["exam_id"].strip()
        student_id = row["student_id"].strip()
        reason = row["reason"].strip()

        if not exam_id or not student_id or not reason:
            print("Skipped a record with missing information.")
            invalid_count += 1
            continue

        # Check whether the exam exists
        if not find_exam(exam_id):
            print(f"Unknown exam ID: {exam_id}")
            invalid_count += 1
            continue

        # Check whether the student exists
        if student_id not in students:
            print(f"Unknown student ID: {student_id}")
            invalid_count += 1
            continue

        valid_records.setdefault(exam_id, {})

        if student_id in valid_records[exam_id]:
            print(
                f"Duplicate detention record: "
                f"{student_id} in exam {exam_id}"
            )
            invalid_count += 1
            continue

        valid_records[exam_id][student_id] = reason

    detained_students = valid_records

    total = sum(
        len(student_records)
        for student_records in detained_students.values()
    )

    print("\nDetained students file loaded.")
    print("Valid detention records:", total)
    print("Invalid or duplicate records:", invalid_count)
    print("Exams with detention records:", len(detained_students))
```

```python
def view_detained_students():
    if not detained_students:
        print("No detained students found.")
        return

    for exam_id, student_records in detained_students.items():
        exam = find_exam(exam_id)

        print("\n========== DETAINED STUDENTS REPORT ==========")
        print("Exam ID:", exam_id)

        if exam:
            print("Subject:", exam.get("subject", "N/A"))
            print("Date:", exam.get("date", "N/A"))
            print("Time:", exam.get("time", "N/A"))

        print("-" * 75)
        print(f"{'Student ID':<15}{'Name':<25}{'Roll No.':<15}Reason")
        print("-" * 75)

        for student_id, reason in student_records.items():
            student = students[student_id]

            print(
                f"{student_id:<15}"
                f"{student.get('name', 'N/A'):<25}"
                f"{student.get('roll_no', 'N/A'):<15}"
                f"{reason}"
            )

        print("-" * 75)
        print("Total detained students:", len(student_records))
```
